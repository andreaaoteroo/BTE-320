```python
number_one = float(input('Please enter your first number: '))
number_two = float(input('Please enter your second number: '))
symbol = input('Please enter your symbol: ')
if symbol == '+':
  print(f'{number_one} + {number_two} = {number_one + number_two}')
elif symbol == '-':
  print(f'{number_one} - {number_two} = {number_one - number_two}')
elif symbol == '*':
  print(f'{number_one} * {number_two} = {number_one * number_two}')
elif symbol == '/':
  print(f'{number_one} / {number_two} = {number_one / number_two}')
elif symbol == '//':
  print(f'{number_one} // {number_two} = {number_one // number_two}')
elif symbol == '**':
  print(f'{number_one} ** {number_two} = {number_one ** number_two}')
elif symbol == '%':
  print(f'{number_one} % {number_two} = {number_one % number_two}')
else:
  print('Invalid Inputs')
```

```python
import random
play = input('Would you like to play the game? (yes/no): ')
while play == 'yes':
  P1 = input('P1 choose between rock/paper/scissors: ')
  P2 = random.choice(['rock', 'paper', 'scissors']).capitalize()

  print(f'Player chose {P1}; Computer chose {P2}')

  if (P1 == 'rock' and P2 == 'scissors') or (P1 == 'paper' and P2 == 'rock') or (P1 == 'scissors' and P2 == 'paper'):
    print('Player wins')
  elif P1 == P2:
    print('Draw')
  else:
    print('Computer wins')
  play = input('Would you like to play the game? (yes/no): ')
print('Good day!')
```

```python
def reverse(s):
  if len(s) <= 1:
    return s
  else:
    return s[-1] + reverse(s[1:-1]) + s[0]

reverse('Hello')
```

```python
def calcFee(hours):
  fee = 5 + 2.5 * hours
  if fee < 10.0:
    fee = 10.0
  elif fee >= 20.0:
    fee = 20.0
  print(f'Fee: ${fee}')
def main():
  try:
    numHours = int(input('Enter number of hours: '))
    calcFee(numHours)
  except ValueError:
    print('Cannot cast to int().')
  except:
    print('Unkown error...')
if __name__ == '__main__':
  main()
```

```python
def getValues(n):
  L = []

  for i in range(n):
    weight = float(input('Enter weight: '))
    L.append(weight)

  return L

def calcAverage(alist):
  try:
    print(f'The average weight is {sum(alist[2:]) / len(alist[2:])} lbs.')
  except ZeroDivisionError:
    print('List alist is empty!')

if __name__ == '__main__':
  try:
    n = int(input('Enter size of list: '))
    alist = getValues(n)
    calcAverage(alist)
  except ValueError:
    print('Cannot cast to int().')
```

```python
class Greeting:
  def __init__(self, first, last):
    self.first = first
    self.last = last
  def display(self, formal = False):
    if formal:
      print(f'Hi, {self.first} {self.last}')
    else:
      print(f'Hi {self.first}')
  def __str__(self):
    return f'{self.first}, {self.last}'

greeting = Greeting('Donald', 'Becker')
greeting.display(formal = True)
print(greeting)
```
